```python
import numpy as np
import matplotlib.pyplot as plt

# 数据
x_labels = ['4', '8', '12', '16', '32']
methods = ['RTSS2020','sen+old','sen+new','HEFT+new','CPOP','OUR']
'''data = [[0.99,0.989,0.981,0.979,0.985,0.971],#10
[0.979,0.973,0.966,0.953,0.98,0.927],#39
[0.973,0.967,0.958,0.936,0.977,0.905],#53
[0.967,0.957,0.953,0.914,0.977,0.882],#71
[0.939,0.924,0.919,0.88,0.984,0.848]]#71'''
data = [[0.998,0.92,0.919,0.914,0.913,0.912,0.901],#10
[1,0.797,0.794,0.783,0.781,0.782,0.75],#39
[1,0.646,0.641,0.632,0.628,0.631,0.59],#53
[1,0.592,0.586,0.575,0.567,0.581,0.528],#71
[1,0.424,0.416,0.411,0.402,0.424,0.369]]#71
# 遍历每一行
for row in data:
    # 获取第二列的元素
    second_element = row[3]
    
    # 对该行的每个元素进行操作
    for i in range(len(row)):
        # 减去第二列的元素，并除以第二列的元素
        row[i] = (second_element - row[i]) / second_element
    
    # 删除第二列
    del row[3]
data = np.array(data)#5*6

# 设置柱状图参数
bar_width = 0.15
x = np.arange(len(x_labels))
colors = ['r', 'g', 'b', 'c', 'm', 'y']

# 绘制并列柱状图
fig, ax = plt.subplots(figsize=(20,10))
for i in range(len(methods)):
    ax.bar(x + i * bar_width, data[:, i], width=bar_width, label=methods[i])

# 设置图例和标签
ax.set_xticks(x + (len(methods) - 1) * bar_width / 2)
ax.set_xticklabels(x_labels)
ax.legend()

# 设置x轴标签和标题
plt.xlabel('cores', fontsize=20)
plt.ylabel('avg_makespan_percent', fontsize=20)
plt.title('Differ cores', fontsize=20)
plt.ylim(-1.5,0.2)

# 显示图形
plt.show()
```
