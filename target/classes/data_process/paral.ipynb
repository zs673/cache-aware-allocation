```python
import numpy as np
import matplotlib.pyplot as plt

# 数据
x_labels = ['10', '15', '20', '25', '30']
methods = ['RTSS2020','sen+old','sen+new','HEFT+new','CPOP','OUR']
'''data = [[0.93,0.924,0.912,0.887,0.99,0.866],#46
[0.95,0.939,0.94,0.903,0.99,0.872],#68
[0.97,0.954,0.95,0.911,0.985,0.878],#72--20
[0.972,0.965,0.949,0.915,0.98,0.88],#61
[0.987,0.976,0.962,0.939,0.975,0.906]]#56'''
data = [[1,0.535,0.53,0.521,0.512,0.563,0.493],#46
[1,0.555,0.549,0.541,0.529,0.57,0.505],#68
[1,0.566,0.559,0.55,0.539,0.57,0.509],#72--20
[1,0.576,0.569,0.558,0.549,0.571,0.516],#61
[1,0.592,0.586,0.575,0.567,0.581,0.528]]#56
# 遍历每一行
for row in data:
    # 获取第二列的元素
    second_element = row[3]
    
    # 对该行的每个元素进行操作
    for i in range(len(row)):
        # 减去第二列的元素，并除以第二列的元素
        row[i] = (second_element - row[i]) / second_element
    
    # 删除第二列
    del row[3]
data = np.array(data)#5*6

# 设置柱状图参数
bar_width = 0.15
x = np.arange(len(x_labels))
colors = ['r', 'g', 'b', 'c', 'm', 'y']

# 绘制并列柱状图
fig, ax = plt.subplots(figsize=(20,10))
for i in range(len(methods)):
    ax.bar(x + i * bar_width, data[:, i], width=bar_width, label=methods[i])

# 设置图例和标签
ax.set_xticks(x + (len(methods) - 1) * bar_width / 2)
ax.set_xticklabels(x_labels)
ax.legend()

# 设置x轴标签和标题
plt.xlabel('maxparal', fontsize=20)
plt.ylabel('avg_makespan_percent', fontsize=20)
plt.title('Differ maxparal', fontsize=20)
plt.ylim(-1,0.1)

# 显示图形
plt.show()
```
