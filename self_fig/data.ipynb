```python
import numpy as np
import pandas as pd
import os
```

```python
# 读取文本文件
file_abr = "./2L1/A_makespan_abs_1_"

# 指定要读取的行范围，例如从第1行到第5行
start_line = 507# 5 507
end_line = 1006# 504 1006

data = []

for i in range(4, 40, 4):
    filename = file_abr + str(i / 10) + ".txt"
    with open(filename, "r") as file:
        lines = file.readlines()
        for line in lines[start_line - 1:end_line]:
            # 以逗号分割每一行的数据，并转换成浮点数
            values = [float(val) for val in line.strip().rstrip(',').split(",")]
            values.insert(0, i / 10)
            data.append(values)

# 创建一个DataFrame对象
df = pd.DataFrame(data, columns=["util","avg", "med", "max", "min"])

# 将数据写入Excel文件
excel_filename = "2L1_AJLR.xlsx"
df.to_excel(excel_filename, index=False)
```
