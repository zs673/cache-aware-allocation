```python
import numpy as np
import matplotlib.pyplot as plt

# 数据
x_labels = ['4', '8', '12', '16', '32']
methods = ['sen+old','sen+new','HEFT','HEFT+new','CPOP','OUR']
data = [[0.986,0.975,0.979,0.966,0.987,0.965],
[0.974,0.964,0.959,0.941,0.984,0.937],
[0.963,0.946,0.946,0.916,0.985,0.914],
[0.947,0.926,0.923,0.89,0.985,0.883],
[0.886,0.868,0.868,0.829,0.989,0.827]]
data = np.array(data)#5*6

# 设置柱状图参数
bar_width = 0.15
x = np.arange(len(x_labels))
colors = ['r', 'g', 'b', 'c', 'm', 'y']

# 绘制并列柱状图
fig, ax = plt.subplots(figsize=(20,10))
for i in range(len(methods)):
    ax.bar(x + i * bar_width, data[:, i], width=bar_width, label=methods[i])

# 设置图例和标签
ax.set_xticks(x + (len(methods) - 1) * bar_width / 2)
ax.set_xticklabels(x_labels)
ax.legend()

# 设置x轴标签和标题
plt.xlabel('cores')
plt.ylabel('avg_makespan_percent')
plt.title('Differ cores')

# 显示图形
plt.show()
```
