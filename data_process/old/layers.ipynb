```python
import numpy as np
import matplotlib.pyplot as plt

# 数据
x_labels = ['10', '15', '20', '25', '30']
methods = ['sen+old','sen+new','HEFT','HEFT+new','CPOP','OUR']
data = [[0.939,0.932,0.927,0.899,0.99,0.897],
[0.937,0.928,0.923,0.889,0.991,0.887],
[0.939,0.924,0.921,0.885,0.987,0.884],
[0.935,0.923,0.923,0.889,0.99,0.888],
[0.951,0.929,0.933,0.892,0.988,0.891]]
data = np.array(data)#5*6

# 设置柱状图参数
bar_width = 0.15
x = np.arange(len(x_labels))
colors = ['r', 'g', 'b', 'c', 'm', 'y']

# 绘制并列柱状图
fig, ax = plt.subplots(figsize=(20,10))
for i in range(len(methods)):
    ax.bar(x + i * bar_width, data[:, i], width=bar_width, label=methods[i])

# 设置图例和标签
ax.set_xticks(x + (len(methods) - 1) * bar_width / 2)
ax.set_xticklabels(x_labels)
ax.legend()

# 设置x轴标签和标题
plt.xlabel('maxlayers')
plt.ylabel('avg_makespan_percent')
plt.title('Differ maxlayers')

# 显示图形
plt.show()
```
