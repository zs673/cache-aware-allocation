```python
import numpy as np

filename = 'find/1.1.txt'
group_size = 50
num_groups = 6

data = []
group = []

with open(filename, 'r') as file:
    lines = file.readlines()
    for line in lines:
        line = line.strip()[:5]  # 去除行末的换行符
        if line:  # 如果行不为空
            group.append(float(line))
        elif group:  # 如果行为空且group不为空
            data.append(group)
            group = []
    if group:
        data.append(group)  # 添加最后一组数据

# 将数据转换为numpy.array形式
data_array = np.array(data)

# 打印数据
for i, group_data in enumerate(data_array):
    print(f"Group {i+1}:")
    print(group_data)
    print()
```
